```python
!pip install -q openai

from getpass import getpass
from openai import OpenAI

OPENROUTER_API_KEY = getpass("Enter your OpenRouter API key: ")
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"

OPENAI_MODEL = "openai/gpt-4o-mini"
GEMINI_MODEL = "google/gemini-2.5-flash"
CLAUDE_MODEL = "anthropic/claude-opus-5"


def call_openai(prompt: str) -> str:
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    response = client.chat.completions.create(
        model=OPENAI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""


def call_gemini(prompt: str) -> str:
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    response = client.chat.completions.create(
        model=GEMINI_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""


def call_claude(prompt: str) -> str:
    client = OpenAI(api_key=OPENROUTER_API_KEY, base_url=OPENROUTER_BASE_URL)
    response = client.chat.completions.create(
        model=CLAUDE_MODEL,
        messages=[{"role": "user", "content": prompt}],
        temperature=0.3,
        max_tokens=250,
    )
    return response.choices[0].message.content or ""

print("Setup complete!")
```

```python
CREATIVE_PROMPT = "Write a 4-line poem about learning by testing code."

ANALYTICAL_PROMPT = (
    "Summarize in one sentence (<= 40 words): "
    "'Using multiple LLM providers helps compare creativity, formatting reliability, and speed, "
    "but adds integration overhead and testing complexity.'"
)

INSTRUCTION_PROMPT = (
    "Return ONLY valid JSON (no backticks). The JSON must be an object with key 'todos' "
    "which is a list of EXACTLY 2 items. Each item must be an object with keys "
    "'task' (string) and 'done' (boolean)."
)
```

```python
print("=== CREATIVE WRITING ===")
print("\nOPENAI:\n", call_openai(CREATIVE_PROMPT))
print("\nGEMINI:\n", call_gemini(CREATIVE_PROMPT))
print("\nCLAUDE:\n", call_claude(CREATIVE_PROMPT))

print("\n=== ANALYTICAL REASONING ===")
print("\nOPENAI:\n", call_openai(ANALYTICAL_PROMPT))
print("\nGEMINI:\n", call_gemini(ANALYTICAL_PROMPT))
print("\nCLAUDE:\n", call_claude(ANALYTICAL_PROMPT))

print("\n=== INSTRUCTION FOLLOWING ===")
print("\nOPENAI:\n", call_openai(INSTRUCTION_PROMPT))
print("\nGEMINI:\n", call_gemini(INSTRUCTION_PROMPT))
print("\nCLAUDE:\n", call_claude(INSTRUCTION_PROMPT))
```
